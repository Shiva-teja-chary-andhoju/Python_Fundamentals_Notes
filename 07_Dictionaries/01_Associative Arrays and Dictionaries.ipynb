# Associative Arrays and Dictionaries

## Associating Things

➔ ASCII table

  ➔ associates a numeric value to certain characters

```text
A  →  65          a  →  97          space → 32
B  →  66          b  →  98          <     → 60

...

Z  →  90          z  → 122          @     → 64
                                     ...
```

We could try this:

```python
keys   = [' ', '<', '@', 'A', 'B', ..., 'Z', 'a', 'b', ..., 'z']

values = [32, 60, 64, 65, 66, ..., 90, 97, 98, ..., 122]
```

➔ to find the numerical value of `'A'`

  ➔ `scan` `keys` to find index of `'A'`

  ➔ lookup the value for that index in the `values` list (array)

## Another Approach

instead of storing the data in separate lists, use a list of tuples

→ each tuple has two elements  → `(key, value)` 

```python
items = [('A', 65), … , ('Z', 90), ('a', 97), … , ('z', 122)]
```

→ to find value associated with `'a'`

&emsp;→ scan items looking at `first` item of tuple until we find `'a'`

&emsp;→ the value we want is the `second` element of that tuple we just found

→ both approaches have one major drawback

&emsp;→ must scan an array until we find the correct element

&emsp;→ the longer the array, the longer time this will take (worst case is last element)

## Hash Maps (aka Dictionaries)

→ better implementation is the `hash map` (or `dictionary`)

&emsp;→ similar to the last approach we saw

&emsp;→ but a special mechanism is used to quickly find a key

&emsp;&emsp;&emsp;→ lookup `speed is not affected by size` of dictionary

IMPORTANT  &emsp;&emsp;  → keys must be `hashable`  &emsp;&emsp;  (hence the term hash map)

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;→ what that means exactly is not important now


→ `strings` are hashable  → `numerics` are hashable 

→ `tuples` *may* be hashable (if all the elements are themselves hashable)

→ `lists` are `not` hashable  (in general, `mutable` objects are `not` hashable) 

## Python Dictionaries

→ a dictionary is a data structure that associates a `value` to a `key`

&emsp;→ both `value` and `key` are Python objects

&emsp;→ `key` must be `hashable` type (e.g. `str`, `int`, `bool`, `float`, …) and `unique`

&emsp;→ `value` can be `any type`

&emsp;→ type is `dict`

&emsp;→ it is a collection of `key: value` pairs

&emsp;→ it is `iterable`

&emsp;&emsp;&emsp;→ but it is not a sequence type

&emsp;&emsp;&emsp;→ `values` are looked up by `key`, not by `index`

&emsp;&emsp;&emsp;→ technically there is no ordering in a dictionary

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;(we'll come back to this point!)

&emsp;→ it is a `mutable` collection

## Looking up values in a Dictionary

→ use `[]` just like for sequence types

&emsp;→ but instead of an index value we specify the `key`

```python
d = {
  'a': 97,
  'b': 98,
  'A': 65,
  'B': 66,
  'z': 122,
  'Z': 90
 }
```

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;`d['a']` → `97`

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;`d['Z']` → `90`


## Replacing the Value of an existing Key

```python
d = {
  'symbol': 'AAPL',
  'date': '2020-03-10',
  'close': 285
}
```

To change the value associated to the key `'close'`:

```python
d['close'] = 285.34
```

→ dictionary now looks like this:  

```python
d = {
    'symbol': 'AAPL',
    'date': '2020-03-10',
    'close': 285.34
}
```

## Adding a New `key:value` Pair

→ simply assign a value to a new key

&emsp;&emsp;→ if key exists, it will be `updated` as we just saw

&emsp;&emsp;→ if key does not exist, a new entry is `inserted` with key and value

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;(and this explains why keys in a dictionary are necessarily unique!)

```python
d = {
  'symbol': 'AAPL',
  'date': '2020-03-10',
  'close': 285.34
}
```

<br>

```python
d['open'] = 277.14
```

## →

```python
d = {
    'symbol': 'AAPL',
    'date': '2020-03-10',
    'close': 285.34,
    'open': 277.14
}
```

## Deleting a `key:value` Pair

→ we can remove key:value pairs from a dict

&emsp;&emsp;→ use the `del` keyword

```python
d = {
  'symbol': 'AAPL',
  'date': '2020-03-10',
  'close': 285.34,
  'open': 277.14
}
```

<br>

```python
del d['open']
```

## →

```python
d = {
    'symbol': 'AAPL',
    'date': '2020-03-10',
    'close': 285.34
}
```


## Common Exceptions

certain operations on dictionaries can lead to `KeyError` exceptions

&emsp;&emsp;→ trying to read a non-existent key

&emsp;&emsp;→ trying to delete a non-existent key

trying to use a `non-hashable` object as a `key` leads to a `TypeError` exception

```python
d[[10, 20]] = 100
```

&emsp;&emsp;&emsp;&emsp;&emsp;→ TypeError: `unhashabletype: 'list'`

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;→ `[10, 20]` is a `list`, and lists are not hashable

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;→ `cannot` be used as a key

# Coding

### Associative Arrays and Dictionaries

As we saw in the lecture, dictionaries are essentially a way of associating two values together - a **key** and a **value**.

Think of it like a regular dictionary - the key is the word you are looking for in the dictionary, and the value is the definition of the word.

Dictionaries are organized in a way that makes searching for a word much more efficient than just scanning all the words one by one until you find the one you are looking for.

The same goes with dictionaries in Python, key lookups in dictionaries are very efficient.

We can create a dictionary using a literal:

````python
d = {'a': 1, 'b': 2, 'c': 3}
````

We can also use whitespace to make dictionary literals more readable if we prefer:

````python
person = {
    'first_name': 'Eric',
    'last_name': 'Idle',
    'year_born': 2016 
}
````

To look up the value for a key in a dictionary we use square brackets (just like a sequence, except we don't use indexes, but keys):

````python
person['year_born']
````

To replace the value for an existing key, we just use an assignment:

````python
person['year_born'] = 1943
````

And the data was updated:

````python
person
````

To create a new key in a dictionary, again just use an assignment:

````python
person['month_born'] = 'March'
````

````python
person
````

So far we've been using strings as dictionary keys, and that is very common, but keys can be any Python object that is hashable, like strings, numerics and certain objects (we'll see this later in the course).

````python
d = {3.14: 'pi', 2: 'even', 'prime': 7}
````

````python
d
````

As you can see we have a dictionary with a mix of key types - an integer, a float, and a string.

````python
d[3.14]
````

````python
d[2]
````

````python
d['prime']
````

In general, mutable objects (like a list for example) are not hashable, and so cannot be used as keys in a dictionary.

````python
l = [1, 2, 3]
````

````python
d = {l: 100}
````

If you need to find out whether something is hashable or not, you can use the `hash()` function:

````python
hash(100)
````

````python
hash(l)
````

Tuples are immutable collections, but they may or may not be hashable depending on whether all the contained elements are hashable or not:

````python
t = (1, 2, 3, 4)
````

````python
hash(t)
````

````python
t = ([1, 2], 3, 4)
````

````python
hash(t)
````

You'll notice that the tuple was unhashable because it contained a list, which is, itself, unhashable.

So, we can use tuples as keys sometimes:

````python
d = {
    (0, 0): "origin",
    (1, 0): "unit-x",
    (0, 1): "unit-y"
}
````

````python
d[(0,0)]
````

We can delete a key (and it's associated value) from a dictionary using the `del` keyword, again just like we did with lists:

````python
d = {'a': 1, 'b': 2, 'c': 3}
````

````python
del d['a']
````

You will get exception if you try to lookup or delete a non-existent key:

````python
d['x']
````

````python
del d['x']
````

Later we'll take a look at a dictionary method that can be used to substitute a default value when we look up a non-existent key.

Many many data structures in Python use dictionaries.

For example, all the variables (symbols) we defined in our notebook are actually stored by Python in a dictionary - that's how it is able to quickly access the "value" associated with the variable symbol.

````python
globals()
````

For example, when we access the `person` variable:

````python
person
````

Python actually retrieved the object from the `globals` dictionary:

````python
globals()['person']
````

````python
person is globals()['person']
````

See! Same object.

In Python, dictionaries are everywhere and is probably one of the most important ubiquitous structures in Python.
