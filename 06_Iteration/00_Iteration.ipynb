# Iteration

→ fundamental aspect of writing programs is `repetition`

&emsp;→ want to repeat the same process (code) multiple times

→ how many times?

&emsp;&emsp;→ `known in advance`

&emsp;&emsp;&emsp;&emsp;&emsp;→ load file with 10,000 rows

&emsp;&emsp;&emsp;&emsp;&emsp;→ process each row  → repeat the same process 10,000 times 

&emsp;&emsp;&emsp;&emsp;→ `deterministic`

&emsp;&emsp;→ `not known in advance`

&emsp;&emsp;&emsp;&emsp;&emsp;→ get commodity tick data

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;→ analyze data until ask price falls below some level

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;→ then do something else and stop processing

&emsp;&emsp;&emsp;&emsp;&emsp;→ process may repeat 10 times, or 100 times, we don't know in advance

&emsp;&emsp;&emsp;&emsp;→ `non deterministic`

→ this repetition  is called `iteration`

**deterministic iteration**

&emsp;&emsp;→ we iterate over the elements of some container

&emsp;&emsp;&emsp;&emsp;→ e.g. sequences

&emsp;&emsp;&emsp;&emsp;→ more generally over objects that are `iterable`

&emsp;&emsp;&emsp;&emsp;→ not all iterables are sequences

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;→a bag of marbles is iterable, but it is not a sequence!

&emsp;&emsp;→ `for` loop

**non-deterministic iteration**

&emsp;&emsp;→ we iterate while some condition is `True`

&emsp;&emsp;→ `while` loop
