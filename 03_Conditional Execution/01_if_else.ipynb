# Conditional Execution

→ one of the fundamental constructs in programming is conditional execution

&emsp;&emsp;→ if something is true

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;→ run some code

&emsp;&emsp;→ else (optionally)

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;→ run some other piece of code

For example, for an ATM withdrawal:

→ if amount does not exceed available funds and does not exceed daily limit

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;→ dispense cash

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;→ print receipt

→ otherwise

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;→ deny request

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;→ display some text on screen

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;→ print slip containing reason

→ this is the primary reason we studied conditional expressions in the last chapter!

# if… else…

## The `if` Statement


```text

                            note the colon!
                                 ↑
if <expression evaluates to True>:    ⎫
    code line 1                       ⎪ notice how this code block is indented
    code line 2                       ⎪
    ...                               ⎬ → this tells Python that all these lines should
                                      ⎪   be executed if the condition is True
                                      ⎪
                                      ⎭
```

→ you "exit" a code block by unindenting your code

→ Python uses code indentation to group together chunks of code

      → called `code blocks`


→ if you are familiar with other languages such as Java or C/C++,
  this is equivalent to using braces { }

## Examples

```python
price = 200
if price < 250:
    make_purchase()
```

→ the call to `make_purchase()` will only be executed if `price < 250` evaluates to `True`, which in this case it is

```python
price = 300
if price < 250:
    make_purchase()
```

→ in this case `make_purchase()` is not executed

## Beware!

→ unindenting code from a block, "exits" the block

→ the following is a common mistake

```python
price = 150
if price < 100:
    print('price is below 100, buying…')        } this is the code block
make_purchase()
```

&emsp;→ `price < 100` is `False`

&emsp;&emsp;&emsp;→ does not run code in the `if` block

&emsp;&emsp;&emsp;&emsp;&emsp;(`if` block only contains a **single** statement – the `print` statement)

&emsp;&emsp;&emsp;&emsp;&emsp;→ runs `make_purchase()`  → bug!! 

## The `else` Clause

→ often in conditional execution

&emsp;&emsp;if something is True

&emsp;&emsp;&emsp;→ do something

&emsp;&emsp;otherwise

&emsp;&emsp;&emsp;→ do something else

→ Python's `if` statement supports an `else` clause  → it is optional 

```python
if <expression is True>:
    [Code Block 1]
else:                ←--------
    [Code Block 2]   ←--------
```
&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;←--------note how `else` is `unindented` from the `if` block, and followed by a colon

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;←--------indent to form the `else` block


# Example

```python
price = 200

if price < 250:
    print('The price is right!')
else:
    print('Too pricey!')

print('Done.')
```

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;notice this line of code is unindented

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;→ it has nothing to do with the

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;`if` or `else` blocks

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;→ it will always execute


→ `price < 250` is `True`

→ the `if` block is executed

Output:

```text
The price is right!
```

→ the `else` block is skipped

→ code resumes after the `else` block


Output:

```text
Done.
```

## Nested `if` Statements

→ sometimes we need to nest conditional logic, either in the `if` block or in the `else` block

```python
if price < 1000:
    if price < 500:
        volume = 50
    else:
        volume = 10

    make_purchase(volume)
else:
    print('Too pricey!')
```

Logic:

```text
→ if price < 1000 is True

    → if price < 500
        → set volume to 50

    → otherwise
        → set volume to 10

    → purchase specified volume

→ otherwise
    → Too pricey
```

## Nested `if` Statements

→ the nesting can occur in the `else` block too

```python
if price < 1000:
    make_purchase()
else:
    if price < 2000:
        contact_vendor()
    else:
        find_new_vendor()
```

→ can nest to any number of levels

→ too much nesting can make code hard to read!

→ keep it to a minimum

# Coding

### if...else...

As we saw in the lecure, the `else` clause is optional, so let's start with a plain `if` statement first:

````python
if 1 < 2:
    print('1 is less than 2')
````

````python
if 2 < 1:
    print('2 is less than 1')
````

As you can see, the code block indented under the `if` statement is executed **if and only if** the conditional expression is `True`.

Be careful with code blocks - keep the code indented. If you unindent the code Python will treat this as outside the code block:

````python
if 1 < 2:
    print('block - line 1')
    print('block - line 2')
    print('block - line 3')
print('next line')
````

As you can see all the `print` statement executed, but what actually happened is that `next line` would always execute because it is **not** in the `if` block

````python
if 2 < 1:
    print('block - line 1')
    print('block - line 2')
    print('block - line 3')
print('next line')
````

See, `next line` still printed. A code block is identified by indenting it one level more than the previous line of code, and is terminated by unindenting you next line of code. Most editors, including Jupyter allow you to use the `Tab` key to indent, and `Shift + Tab` to unindent.

To make things clearer, we often add a blank line after the end of a block:

````python
if 2 < 1:
    print('block - line 1')
    print('block - line 2')
    print('block - line 3')
    
print('next line')
````

This makes the code a bit easier to read (easier to sperate the code block from the next line).

Code blocks are used frequently throughout Python, so we'll see this many times over.

Next, let's look at the `else` clause. 

The code block under the `else` clause will execute **if and only if** the condition is `False` (in which case the `if` code block does not execute).

In other words, the code blocks under the `if` and under the `else` are mutually exclusive - if one runs, the other does not, and vice versa.

````python
if 1 < 2:
    print('1 is less than 2')
else:
    print('1 is not less than 2')
````

````python
if 2 < 1:
    print('2 is less than 1')
else:
    print('2 is not less than 1')
````

Of course our conditional expression in the `if` statement can be more complex:

````python
account_enabled = True
balance = 1000
withdraw = 100

if account_enabled and withdraw <= balance:
    print('withdrawal authorized')
else:
    print('withdrawal not authorized')
````

````python
account_enabled = True
balance = 1000
withdraw = 10_000

if account_enabled and withdraw <= balance:
    print('withdrawal authorized')
else:
    print('withdrawal not authorized')
````

````python
account_enabled = False
balance = 1000
withdraw = 100

if account_enabled and withdraw <= balance:
    print('withdrawal authorized')
else:
    print('withdrawal not authorized')
````

Code blocks can contain any valid Python code.

Amongst other things, they can contain other `if...else...` statements too!

For example, in the account example we did above, we may want to provide a better error message that is more specific - like "account disabled" or "insufficient funds".

We could do it this way:

````python
account_enabled = False
balance = 1000
withdraw = 100

if account_enabled and withdraw <= balance:
    print('withdrawal authorized')
else:
    # cannot withdraw for some reason
    if not account_enabled:
        print('account disabled')
    else:
        # must be insufficient funds
        print('insufficient funds')
````

````python
account_enabled = True
balance = 1000
withdraw = 100_000

if account_enabled and withdraw <= balance:
    print('withdrawal authorized')
else:
    # cannot withdraw for some reason
    if not account_enabled:
        print('account disabled')
    else:
        # must be insufficient funds
        print('insufficient funds')
````

Too many levels of nested code can make it harder to read - so try to avoid this if you can.

For example, suppose we need to assign a letter grade based on a range of numerical grades:
```
>= 90       --> A
>= 80, < 90 --> B
>= 70, < 80 --> C
>= 60, < 70 --> D
< 60        --> F
```

We could write it this way:

````python
grade = 72

if grade >= 90:
    letter_grade = 'A'
else:
    if grade >= 80:
        letter_grade = 'B'
    else:
        if grade >= 70:
            letter_grade = 'C'
        else:
            if grade >= 60:
                letter_grade = 'D'
            else:
                letter_grade = 'F'
                
print(letter_grade)
````

This can get difficult to read.
A slightly different approach could be as follows:

````python
grade = 72

if grade >= 90:
    letter_grade = 'A'

if grade >= 80 and grade < 90:
    letter_grade = 'B'
    
if grade >= 70 and grade < 80:
    letter_grade = 'C'
    
if grade >= 60 and grade < 70:
    letter_grade = 'D'
    
if grade < 60:
    letter_grade = 'F'
    
print(letter_grade)
````

This works, but a lot of needless comparisons - that's not ideal either.

Also there is a danger that all the `if` statement in our code do not account for all possible grades - they do in this case, but what if we had forgottn one interval, or gotten some of the intervals wrong?

We could try this approach too, where we assign a default grade as the lowest grade (so we are ensured we always have a grade), and that results in less comparisons:

````python
grade = 72

letter_grade = 'F'

if grade >= 60:
    letter_grade = 'D'
    
if grade >= 70:
    letter_grade = 'C'
    
if grade >= 80:
    letter_grade = 'B'
    
if grade >= 90:
    letter_grade = 'A'
    
print(letter_grade)
````

So less comparisons, and we are at least guarantee for `letter_grade` to have a value (but maybe the wrong one if we missed an interval or forgot one altogether) - so that's a bit better than the last approach in those respects.

But we've also done a lot of unecessary work!

To assign the grade `C` we:
- first assigned 'F'
- then assigned 'D'
- then assigned 'C'
- then performed the `grade >= 80` check but did not assign a new grade
- then performed the `grade >= 90` check but did not assign a new grade

So this is far from ideal either!
