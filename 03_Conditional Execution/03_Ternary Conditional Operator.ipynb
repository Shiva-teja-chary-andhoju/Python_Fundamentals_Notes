## The conditional ternary operator

→ remember that an operator operates on operands and `returns` (calculates) some `result`

```python
if <conditional exp>:
    var = value1
else:
    var = value2
```

→ in this case we want the ternary operator's operands to be:

&emsp;&emsp;→ the conditional expression

&emsp;&emsp;→ the value to return if the expression is `True`

&emsp;&emsp;→ the value to return if the expression is `False`

## The conditional ternary operator

```python
if <conditional exp>:
    var = value1
else:
    var = value2
```

```python
value1 if<conditional exp> elsevalue2
```

&emsp;→ this is a `single` ternary operator

&emsp;→ if condition is `True`, it returns `value1`

&emsp;→ if condition is `False`, it returns `value2`

## Example

```python
if price < 100:
    volume = 10
else:
    volume = 1
```

→ can be re-written using a conditional ternary operator

```python
volume = 10 ifprice < 100 else1
```

## General Form

→ we saw examples where we used values as the return operands

&emsp;→ but it's more general than that

&emsp;→ the two value operands can be any expression

&emsp;→ the result of the expression is then used

```python
<exp1> if <condition> else <exp2>
```

```python
var = (a –b) if a > b else (b –a)
```

## Short-Circuiting

Just like we saw with Boolean operators, the ternary operator also uses short-circuit evaluation

```python
<exp1> if <condition> else <exp2>
```

&emsp;&emsp;&emsp;&emsp;&emsp;→ first evaluates `<condition>`

&emsp;&emsp;&emsp;&emsp;&emsp;→ if it is `True`, evaluates and returns `<exp1>`

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;→ but does not evaluate `<exp2>`

&emsp;&emsp;&emsp;&emsp;&emsp;→ if it is `False`, evaluates and returns `<exp2>`

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;→ but does not evaluate `<exp1>`

## Example

```python
result = a / b if b != 0 else 'NaN'
```

a = 10&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;→ returns `2`

b = 5&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;→ `b` is `5`, so `b != 0` evaluates to `True`

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;→ `a / b` is calculated and returned


a = 10&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;→ this works just fine, and returns `NaN`

b = 0&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;→ `b` is zero, so `b != 0` evaluates to `False`

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;→ `a / b` is **not** calculated

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;(thereby avoiding a division by zero exception)

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;→ `NaN` is returned

Let's see how the ternary conditional operator works.

Suppose we want to calculate the volume of a bid, based on the ask price of some security.

We could do it this way:

````python
ask_price = 100

if ask_price > 50:
    volume = 50
else:
    volume = 80
    
print(volume)
````

But we can simplify this, without losing readability, as follows:

````python
ask_price = 100

volume = 50 if ask_price > 50 else 80

print(volume)
````

Of course, the "return value" can be any expression, not just a literal value:

Let's say we want to find the "distance" between two numbers, but we forgot about the `abs()` function.

We could do it this way:

````python
a = 10
b = 20

distance = a - b if a >= b else b-a

print(distance)
````

````python
a = 20
b = 10

distance = a - b if a >= b else b-a

print(distance)
````

This ternary operator can also be very useful to deall with "null" values.

Often when we process data, we might have some missing data that we want to replace with either a default value or some calculated value (maybe from the previous row, or the average of all the previous rows, etc).

Suppose we are looping (iterating) through some data file and for every row we process we have the following data:

- `current_value` - the value of some column in our current row. A value of `-999` indicates the value is missing.
- `running_total` - running total of that column's values so far, possibly zero
- `running_count` - running count of number of rows (not including current one)

So, at some point in time we may have this data:

````python
current_value = 100
running_total = 15000
running_count = 125
````

We then calculate the `cleaned_value` this way before we update `running_total` (and updating `running_count` would be done similarly):

````python
cleaned_value = 0 if current_value == -999 else current_value
running_total = running_total + cleaned_value

print(running_total)
````

As you can see, our value was not missing, so we used it as is.

On the other hand, if the value was missing, we might have data like this:

````python
current_value = -999
running_total = 15000
running_count = 125
````

And the same code as before would yield:

````python
cleaned_value = 0 if current_value == -999 else current_value
running_total = running_total + cleaned_value

print(running_total)
````

If we wanted to avoid that `cleaned_value` temporary variable, we could re-write our code this way:

````python
running_total = running_total + (0 if current_value == -999 else current_value)
print(running_total)
````

Be careful here! the ternary conditional operator has **lower** precedence than `+`, so we **need** to use parentheses to make our code behave the way we want (first calculate the cleaned up value, and then add it to the running total)

Keep the ternary operator in your backpocket for **simple** use cases. If it starts to get convoluted, use a regular `if...elif...else` statement!
