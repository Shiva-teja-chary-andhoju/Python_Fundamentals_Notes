# Objects

# What are objects?

→ entities created by Python

$$
\left.
\begin{aligned}
&\rightarrow\ \text{they have state (data)} \\[12pt]
&\rightarrow\ \text{they have methods (functionality)}
\end{aligned}
\right\}
\quad \text{encapsulation}
$$

→ they often represent real world things

## Car

### State (data)

- brand → Toyota
- model → Prius LE
- doors → 4
- model_year → 2020
- odometer → 5,402

### Functionality

- accelerate()
- brake()
- set_cruise_control()
- left_turn_signal_on()

## Integers are Objects

→ an `int` is an object

&emsp;→ it has `state` – the `value` of the integer

&emsp;→ but it also has `functionality`

&emsp;&emsp;&emsp;→ knows how to add itself to another integer

&emsp;&emsp;&emsp;&emsp;&emsp;`(10).__add__(100)` → `110`

&emsp;&emsp;&emsp;&emsp;&emsp;→ an integer object has the method `__add__` used to implement addition

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;(this is not how we typically add two integers together – but that's just syntax)

&emsp;&emsp;&emsp;→ knows how to represent itself as a string (e.g. for visual output)


## Floats are Objects

→ `float` numbers are objects too

&emsp;&emsp;state → `value`

&emsp;&emsp;functionality → `__add__`

&emsp;&emsp;other functionality too, for example:

```python
(0.125).as_integer_ratio()
```

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;→ `1, 8`

## Everything in Python is an object

→ any data type we have in Python is actually an object

$$
\left.
\begin{aligned}
&\rightarrow \text{ it has state} \\
&\rightarrow \text{ it has functionality}
\end{aligned}
\right\}
\text{ attributes}
$$

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;attributes encompass state and functionality

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;some attributes are for state

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;some attributes are for functionality

## Dot Notation

If an object has attributes, how do we access those attributes?

&emsp;→ **dot notation**

&emsp;`car.brand` → accesses the `brand` attribute of the `car` object

&emsp;`car.model` → accesses the `model` attribute of the `car` object

For attributes that represent **functionality**, we usually have to **call** the

attribute to **perform the action**

&emsp;&emsp;&emsp;→ often supplying additional parameters

```python
car.accelerate(10, "mph")
(10).__add__(100)
```

## Mutability and Immutability

→ an object is `mutable` if its internal state `can` be changed

&nbsp;&nbsp;&nbsp;&nbsp;→ one or more data attributes can be changed

<br>

→ an object is `immutable` if its internal state `cannot` be changed

&nbsp;&nbsp;&nbsp;&nbsp;→ the state of the object is "set in stone"

<br>

In Python many data types are immutable:

- integers
- floats
- booleans
- strings
- ...

<br>

While some are mutable:

- lists
- dictionaries
- sets
- ...

# Coding

### Objects

We'll cover this later in this course, but let me first show you how we can create custom objects with state and functionality.

So, don't worry if you don't understand this code right now, you soon will!

````python
class Account:
    def __init__(self, account_number, account_type, initial_balance):
        self.account_number = account_number
        self.account_type = account_type
        self.balance = initial_balance
        
    def deposit(self, amount):
        # should also check that amount is a numerical value!
        if amount > 0:
            self.balance = self.balance + amount
            print(f'Deposited {amount}')
            print(f'New balance is: {self.balance}')
        else:
            print(f'{amount} is an invalid amount.')
            
    def withdraw(self, amount):
        # should also check that amount is a numerical value!
        if amount > 0 and amount <= self.balance:
            self.balance = self.balance - amount
            print(f'Withdrawal: {amount}')
            print(f'New Balance: {self.balance}')
        else:
            if amount < 0:
                print(f'{amount} is an invalid amount')
            else:
                print('Insufficient funds.')
                print(f'Current balance is {self.balance}')
````

````python
my_account = Account('123-456', 'savings', 1_000.00)
````

`my_account` is an **object** with **state**:

````python
my_account.account_number
````

````python
my_account.account_type
````

````python
my_account.balance
````

And it has **functionality**:

````python
my_account.deposit(100)
````

````python
my_account.withdraw(600)
````

````python
my_account.withdraw(10_000)
````

As you can see we can access state and functionality of objects using this **dot** notation.

In Python, everything is an object - this means everything we work with has state and functionality.

For example, integers are objects - they have **state** (their value), as well as **functionality** (they know how to add another number to themselves):

````python
10 + 15
````

The `+` operator is actually using a functional attribute of the integer, called `__add__`.

We could have done the addition this way:

````python
(10).__add__(15)
````

In the same way, `floats` are also objects:

````python
0.125
````

One piece of functionality they have is the `as_integer_ratio` method, which can be useful if you want to get an exact representation of the float:

````python
(0.125).as_integer_ratio()
````

As you can see `0.125` is stored **exactly** as `1/8`.

But as we know, not every float literal we specify can be stored exactly as we wrote it:

````python
0.1
````

This may look like Python is storing this as `0.1`, but we know better:

````python
format(0.1, '.25f')
````

But what is the **exact** number?

````python
(0.1).as_integer_ratio()
````

As we can see it is some big fraction, very close to `1`, but not exactly `1`.

We'll come back to these kinds of topics, but for now the main take away is that everything we work with in Python is an object.
