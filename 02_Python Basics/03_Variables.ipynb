# Variables

## Naming Objects

We often need to `label` objects with some `name`

&emsp;&emsp;→ reminds us what the object is used for

`apy`

`account_balance`


&emsp;&emsp;→ allows us to use the same object in multiple parts of our code

# Assigning Names

To assign a label to an object we use the `assignment` operator `=`

```python
account_balance = 1000.0

apy = 0.25
```

`=` is **not** the mathematical equality symbol


&nbsp;&nbsp;&nbsp;&nbsp;→ we are `assigning` the label `apy` to the object `0.25`


&nbsp;&nbsp;&nbsp;&nbsp;→ we say `apy` is a `reference` to the float object `0.25`

→ the symbol `apy` is just a `label` currently pointing to (or `referencing`) the object `0.25`

## References and Variables

Another way of looking at this:

<br>

```text
             references
             the object
apr  ------------------------->  +---------+
↑                                |  float  | <---Same object in the memory
|                                |         |
a label                          |  0.25   |
                                 +---------+
```

```python
apr = 0.25
```

→ `apr` is called a `variable`

→ but it is just a label (a symbol) that references some object in memory

## Variables

So why the term `variable`?

&nbsp;&nbsp;&nbsp;&nbsp;→ over time, which object a symbol references can change

```python
a = 100
```

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;→ `a` is referencing the object `100`

later in the program...

```python
a = True
```

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;→ `a` is now referencing the object `True`

→ the state of the object the symbol references can change (mutate)

<br>

```text
      a
      |
      |----------------->  +---------------+                           
                           |     list      | 
                           |  1, 2, 3, 4   | ← we append an element to the list
                           +---------------+
```

<br>

→ `a` is still referencing the same object, but the object's **state** has changed (**mutated**)

# How Variable Assignment Happens

```python
apy = 0.25
```
LHS = RHS

<br>

→ Python evaluates the RHS first

&nbsp;&nbsp;&nbsp;&nbsp;→ then it "assigns" that result to the symbol in the LHS

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;(the LHS becomes a named reference to whatever results from the RHS)

<br>

Generally, RHS could be a more complex expression than just a literal

```python
balance = 1000.0 - 50.0

circ = 2 * 3.14 * 1.5
```

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;→ in both cases, the RHS is fully evaluated first

## Using Variables

Once a variable has been created, it can be used elsewhere in the program

```python
pi = 3.1415

radius = 1

circ = 2 * pi * radius
```

```text
pi      ───────────────→ [ 3.1415 ]

radius  ───────────────→ [ 1 ]

circ    ───────────────→ [ 6.283 ]
```

→ `circ` is now a reference to the float `6.283`

```python
radius = 2
```

```text
radius  ───────────────→ [ 2 ]

circ    ───────────────→ [ 6.283 ]
```

BUT this does not change `circ`

&nbsp;&nbsp;&nbsp;&nbsp;→ it still points to `6.283`

# Variable Naming

→ case sensitive

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;`apr` is a different symbol than `APR`

&nbsp;&nbsp;&nbsp;&nbsp;→ **must** follow certain rules

&nbsp;&nbsp;&nbsp;&nbsp;→ **should** follow certain conventions

# Must-Follow Rules

**start** with underscore (`_`) or letter (`a-z`, `A-Z`)

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;(unicode characters are actually OK, but stick to `a-z` `A-Z`)

<br>

**followed** by any number of underscores or letters, or digits (`0-9`)

```text
var      my_var      index1      index_1 ⎫
                                         ⎬ all legal
_var     __var       __add__             ⎭
```
<br>

→ cannot be reserved words

```text
True   False   if   def   and   or
```

and many more we'll come across in this course

## Should-Follow Conventions

PEP 8 Style Guide → typical conventions followed by most Python devs

https://www.python.org/dev/peps/pep-0008/

terminology:

### camel case

→ separate words are distinguished by upper case letters

```python
accountBalance
BankAccount
```

### snake case

→ separate words are distinguished by underscores

```python
account_balance
bank_account
```

## Should-Follow Conventions

For standard variables:

&emsp;&emsp;→ snake case

&emsp;&emsp;→ all lower case letters

```python
account_balance ✅

```

→ Good idea to follow standard conventions

&emsp;&emsp;→ but sometimes you may want to break those conventions

# Coding

### Variables

Recall that we use the `=` symbol as the **assignment** operator. 

We specify a variable name (a label) on the left hand side of the operator, and the value (or expression that evaluates to a value) on the right hand side.

````python
a = 100
````

Now `a` is a symbol (a variable) in our program that we can recall:

````python
a
````

We can use it in subsequent parts of our code:

````python
b = a + 11
````

````python
b
````

As you can see, here we performed a calculation in the right hand side, and assigned that resulting value to the label on the left (`b`).

Note that the `=` symbol is not the same as the mathematical *equality* symbol, so something like this is not valid Python, and we'll get an exception:

````python
10 = 10
````

Note also that the object the symbol `a` or `b` points to can change over time:

````python
a
````

````python
a = 3.14
````

````python
a
````

As you can see, `a` used to reference the integer `100`, but now it references a different object, the float `3.14`.

#### Naming Rules

As discussed in the lecture, there are certain rules you must follow when choosing variable names.

For example, all the following are valid variable names:

````python
test = 10
test_1 = 10
_test_1_ = 10
__test__ = 10
TEST = 10
````

But a variable name may not start with a digit:

````python
1_test = 10
````

And we should not use reserved keywords or built-in functions.

For example, `if` is a reserved keyword:

````python
if = 10
````

There are other symbols that Python has pre-defined for us and although we can change them, we should not.

For example, `float` is used by Python to represent the float data type.

In fact, we can create a float this way:

````python
a = float(10)
````

````python
a
````

But, we *could* remap the symbol `float` to be something else:

````python
float = 100.5
````

The problem now is that the original definition of `float` has been lost to us (internally Python still has it, but **we** don't!)

````python
a = float(10)
````

Since I don't want to "overwrite" Python's symbol for `float`, I am going to delete that symbol I created:

````python
del float
````

Now **my** definition of `float` is gone, and the **original** one provided by Python is available again.

````python
a = float(10)
print(a)
````

#### Naming Conventions

Python developers typically follow a standard naming convention for variable names. Depending on the type of variable, there are different conventions.

In particular, we follow the PEP8 conventions:

https://www.python.org/dev/peps/pep-0008/

You should try reading this guide once in a while (not all in one sitting, but just here and there as time permits - it's quite useful).

For now, we are dealing with simple variables in our code, so we follow the snake case convention, with all letters lower cased:

````python
current_balance = 100.0
````

and not, camel casing like in other languages such as Java or Javascript:

````python
currentBalance = 100.0
````

The other advice I would give is to **always** give your variables meaningful names - not because the computer cares, but because humans reading your code (including yourself six months after you wrote that code) will find it easier to understand what's going on.

Here's an example. Suppose I show you this piece of code, and ask you what does it do:

````python
x = 100
y = 0.1
z = 10

r = x * ((1 + y/12) ** (z * 12))

print(r)
````

How about this code instead, which is identical, except for variable names:

````python
principal = 100
apr = 0.1
years = 10

future_value = principal * ((1 + apr/12) ** (years * 12))

print(future_value)
````
