# Sets

## What are Python sets?

→ just like a mathematical set

&emsp;→ a `collection` of elements

&emsp;→` no ordering` to the elements

&emsp;→ each element is `unique`

→ it is an `iterable`

&emsp;→ but `no guarantee` on what the `iteration order` will be

think of it like a bag of marbles (a `collection` of marbles)

&emsp;&emsp;to iterate you reach in the bag and grab a marble (any marble)

&emsp;&emsp;continue doing so until the bag is empty

&emsp;&emsp;&emsp;&emsp;→ `no order` guaranteed!  → each marble is `unique`! 

Just like mathematical sets, Python supports set `operations`

&emsp;&emsp;&emsp;→ union

&emsp;&emsp;&emsp;→ intersection

&emsp;&emsp;&emsp;→ difference

&emsp;&emsp;&emsp;→ membership (is some object an element of a set or not)

&emsp;&emsp;&emsp;→ containment (subset, strict subset, superset, strict superset)

# Python Sets

think back to `keys` in a dictionary

&emsp;→ they are `unique`

&emsp;→ they are `iterable`

&emsp;→ they have `no particular order` (well, Python 3.6 maintains insertion order)

&emsp;→ keys can be `added or removed` (dictionary is mutable)

&emsp;→ they are hashable too - but leave that aside for a moment

does that remind you of a set?

&emsp;→ we can think of the `keys` in a `dictionary` as a `set`

→ Python's implementation of sets is essentially like a dictionary

&emsp;→ but no values, only `keys`

&emsp;→ because of this, set `elements` must be `hashable` too

## Python Sets

→ type is `set`

→ sets are `iterable`

→ iteration `order` is `not guaranteed` (at least not yet)

→ set `elements` must be `hashable`

→ sets are `mutable`

&emsp;→ `sets are not hashable`

&emsp;&emsp;&emsp;→ a set cannot be an element of another set, or a key in a dictionary

&emsp;&emsp;&emsp;→ if you really want nested sets, use `frozenset`

&emsp;&emsp;&emsp;&emsp;&emsp;→ `immutable` equivalent of sets – those `are hashable`

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;(if all the elements are, themselves, hashable)

## Defining Sets

→ can also make a set from any iterable (of hashable elements)

```python
l = [1, 2, 3, 4, 5]
```

```python
s = set('python')
```

# Coding

Python sets can be considered as a representation of their mathematical equivalent.

But we can think of a set as a dictionary that contains keys only, with no associated value.

And in fact, Python implements sets almost the same way as a dictionary.

Amongst other things, this means that sets in Python can only contain hashable elements (just like dictionary keys have to be hashable). 

The data type for sets is `set`, and we can create sets using literals:

````python
s = {'a', 'b', 'c'}
````

You'll notice that we used curly braces `{}` to define a set literal - which closely resembles what we use for dictionary literals. The difference here is that we have no associated value with the "key":

````python
d = {'a': 1, 'b': 2, 'c': 3}
s = {'a', 'b', 'c'}
````

We can also create a set using the `set()` function:

````python
s = set(['a', 'b', 'c'])
````

````python
s
````

We can use any iterable of hashable elements, so this works too:

````python
s = set('python')
````

````python
s
````

And since set elements are unique, we will not see repeated values in a set:

````python
s = set(['a', 'a', 'b', 'b'])
````

````python
s
````

or

````python
s = set('banana')
````

````python
s
````

Remember how we used a dictionary to find all the unique elements in a string (or any iterable for that matter)?

Instead, we can simply use a set.

To define an empty set we cannot use a literal which would look like this: `{}`

The problem here is that Python would not know whether `{}` means an empty set or an empty dictionary. So Python;s default is to use `{}` as the literal for an empty dictionary (since dictionaries are much more widely used than sets).

To create an empty set, we can just use the `set()` function:

````python
s = set()
````

````python
s
````

````python
len(s)
````

Membership testing, iteration, clearing and copying sets works exactly the same as with dictionaries.

````python
s = set('python')
````

````python
s
````

````python
'p' in s, 'x' not in s
````

````python
for item in s:
    print(item)
````

One thing you have maybe noticed: the iteration order for sets is not guaranteed - unlike Python dictionaries that maintain an insertion order, Python sets do not (not yet anyways!).
