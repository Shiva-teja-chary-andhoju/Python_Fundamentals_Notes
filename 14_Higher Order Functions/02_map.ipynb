# The `map()` Function

→ the `map()` function calls a specified function for every element of some iterable

→ very similar to doing something like this:

```python
def my_map(func, iterable):
    result = [func(element) for element in iterable]
    return result
```

&emsp;→ here we are creating a `list` that contains the function `func` applied to every element of `iterable`

&emsp;&emsp;&emsp;→ but it creates a `list`

&emsp;&emsp;&emsp;→ can take a lot of space if iterable is large

&emsp;&emsp;&emsp;→ especially wasteful if we don't iterate over all the values

→ `map()` returns an iterator

```python
iterator = map(func, iterable)
```

as we iterate over that iterator:

&emsp;&emsp;→ Python moves to the next item in `iterable`

&emsp;&emsp;→ calls `func(element)`

&emsp;&emsp;→ returns the result

&emsp;&emsp;&emsp;&emsp;→ less wasted space

&emsp;&emsp;&emsp;&emsp;→ saves computations if we don't iterate over the whole list

→ equivalently we could also just use a `generator expression`

```python
(func(el) for el in iterable)
```

# Coding

### map

The `map()` function creates an iterator that applies a given function to an iterable, element by element.

Let's see a simple example of this:

````python
data = ['a', 'ab', 'abc', 'abcd']
````

Suppose we want to get the length of each string in `data`.

We could certainly do it this way:

````python
lengths = [len(element) for element in data]
````

````python
lengths
````

One disadvantage here is that we created a list.

If we don't need to iterate over this multiple times (or it's small and computationally cheap), we could use a generator expression instead:

````python
lengths = (len(element) for element in data)
````

````python
list(lengths)
````

But we could also use the `map` function, which offers a slightly cleaner syntax:

````python
lengths = map(len, data)
````

Now the result of `map` is not a list, or a tuple, but an iterator.

````python
lengths
````

We can iterate through it:

````python
list(lengths)
````

But of course, the iterator is now exhausted (this is similar to the `zip` function):

````python
list(lengths)
````

It's important to understand that these kinds of iterators/generators are far more efficient than building a list like we did with the comprehension just now.
