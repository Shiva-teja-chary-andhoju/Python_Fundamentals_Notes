# Virtual Environments

## How does Python run?

→ Installing Python

&nbsp;&nbsp;&nbsp;&nbsp;→ Creates a **directory** on your machine

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;→ Contains **ALL** the files needed to run Python on your OS

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;→ Base Python

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;→ the language itself and a few built-ins (e.g. `for` loops, `zip`, `filter`, etc.)

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;→ Standard Library

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;→ A bunch of additional functionality: e.g. math, stats, data structures, async, threading, and a ton more

→ Python directory contains

&nbsp;&nbsp;&nbsp;&nbsp;→ Binaries and files to run Python (executable code files, like apps basically)

&nbsp;&nbsp;&nbsp;&nbsp;→ Additional code files and executables for standard library

&nbsp;&nbsp;&nbsp;&nbsp;→ When you run Python, it runs the executable in that Python directory

→ You could make a copy of the entire installation folder, and run Python from there

&nbsp;&nbsp;&nbsp;&nbsp;→ In fact, that's how you can have multiple versions of Python installed on your system

&nbsp;&nbsp;&nbsp;&nbsp;→ But even if not dealing with multiple versions of Python there's a reason to make such copies

→ When you install a 3<sup>rd</sup> party library, it basically **adds the necessary files to the Python folder**

→ Suppose you have two Python projects

&nbsp;&nbsp;&nbsp;&nbsp;→ They both use the same version of Python (or may not, but let's say they do)

&nbsp;&nbsp;&nbsp;&nbsp;→ Both projects need a library called `psycopg` (it is used to connect to a Postgres database)

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;→ BUT, project 1 needs version `2` of the library, while project 2 needs version `3` of the library

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;→ Libraries are just files that get copied into your Python directory

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;→ **Cannot install both version 2 and version 3 in the same Python directory!!**

&nbsp;&nbsp;&nbsp;&nbsp;→ solution?

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;→ Make two copies of the base Python installation

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;→ Add version 2 to the first copy, and run Python from there for project 1

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;→ Add version 3 to the second copy, and run Python from there for project 2

## General Approach

→ Always leave your base Python installation clean

&nbsp;&nbsp;&nbsp;&nbsp;→ don't add anything to it

&nbsp;&nbsp;&nbsp;&nbsp;→ keep it as the reference install of Python for your system

→ For each Python project you have

&nbsp;&nbsp;&nbsp;&nbsp;→ Make a **copy** of the reference Python directory

&nbsp;&nbsp;&nbsp;&nbsp;→ Install any additional libraries you need into that copy

&nbsp;&nbsp;&nbsp;&nbsp;→ When you run your Python app, use the Python binary in that copy (which will have access to any additional files (libraries) you installed in it)

→ Doing all this manually can be time consuming and tedious!!

## Virtual Environments

→ A virtual environment is basically a `copy` of Python with any additional libraries you need that you use to run a specific Python project – exactly what we described in the previous slide.

→ There are several tools available to do the tedious work for us

&nbsp;&nbsp;&nbsp;&nbsp;→ `pipenv`  
&nbsp;&nbsp;&nbsp;&nbsp;→ `virtualenv`  
&nbsp;&nbsp;&nbsp;&nbsp;→ Python has one in the standard library too!  
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;→ `venv`

→ These tools allow you to easily **`copy`** your reference Python directory (called a **`virtual environment`**).

→ Install libraries into that copy (virtual env).

→ `activate` the environment

&nbsp;&nbsp;&nbsp;&nbsp;→ Basically, just a fancy way of saying that when you run Python (and your app), it will run the Python binary and all the files from your **copy (virtual env)**, instead of using your reference Python directory.

## 3<sup>rd</sup> Party Libraries

→ many 3<sup>rd</sup> party libraries exist

&emsp;→ add-ons to Python for more specialized functionality

&emsp;&emsp;→ a bunch of files

&emsp;&emsp;→ that get added to your Python "installation"

→ 3<sup>rd</sup> party libraries often rely on other 3<sup>rd</sup> party libraries

&emsp;→ or might have specific releases for specific Python versions

→ this can create conflicts!

| | |
|---|---|
| `my_app_1` → some_lib_1.0 | (`breaks` with some_lib_1.1 and higher) |
| `my_app_2` → some_lib_1.1 | (`breaks` with some_lib_1.0 and lower) |

## Solution

→ since Python is just a directory of files

&emsp;&emsp;→ create two `copies` of this directory

| | |
|---|---|
| /usr/user1/python3.9-ENV1/ | /usr/user1/python3.9-ENV2/ |
| install `some_lib_1.0` in here | install `some_lib_1.1` in here |

&emsp;&emsp;→ run your apps/shell using the appropriate Python directory

&emsp;&emsp;&emsp;→ /Users/user1/python3.9-ENV1/bin/python my_app.py

&emsp;&emsp;&emsp;→ /Users/user1/python3.9-ENV2/bin/python my_other_app.py

→ typing this long path is tedious

→ add to PATH environment variable (tells OS where to look for executables)

## Solution

/usr/user1/python3.9-ENV1/

&emsp;→ add `/usr/user1/python3.9-ENV1/` to (front of) PATH

&emsp;→ `python my_app.py`

/usr/user1/python3.9-ENV2/

→ remove `/usr/user1/python3.9-ENV1/` from path

→ add `/usr/user1/python3.9-ENV2/ `to (front of) PATH

→ `python my_other_app.py`

→ works but can become tedious as well!

- **Versioning Rules**:

  - **Major (2.0.0)**
    - Big changes that might break your code.
    - You should read the release notes before upgrading.
    - Example: A library removes or changes a function you were using.

  - **Minor (2.1.0)**
    - Adds new features, but doesn't break existing ones.
    - Usually safe to upgrade.
    - Example: A new function is added, but old ones still work.

  - **Patch (2.1.3)**
    - Bug fixes or small improvements.
    - Very safe to upgrade.
    - Example: Fixes a typo or a crash in a specific case.

→  **Safety Tips**: Always use venvs to avoid conflicts. Before updating, check library release notes for breaking changes. Update regularly but cautiously to stay secure and current.

##  Important Note: Choose Python Version Based on Library Compatibility
**Golden Rule:** Do not select a Python version first and then try to make old code work. Always check which Python versions are supported by the libraries used in your project.
### Why This Matters
Older projects often depend on specific library versions. Those older library versions may not support newer Python releases.
Example:
```text
Project
│
├── pandas==1.3.5
├── numpy==1.21.6
├── openpyxl==3.0.10
└── Python ?
```
#### Installing Python 3.13 may fail because some older library versions were never built for Python 3.13.

### Recommended Approach for Existing Projects
#### 1. Check Project Dependencies
Look for:
```text
requirements.txt
```
or documentation specifying library versions.
Example:
```text
pandas==1.3.5
numpy==1.21.6
scipy==1.7.3
```

#### 2. Check Python Compatibility
Review the library's supported Python versions.
Example:
```text
numpy 1.21.x
Supports:
Python 3.7 - 3.10
```
Then choose a compatible Python version:
```text
Python 3.10 ✅
Python 3.11 ❌
Python 3.13 ❌
```

#### 3. Install the Compatible Python Version
Example:
```cmd
py -3.10 -m venv myenv
```
instead of:
```cmd
py -3.13 -m venv myenv
```

### Typical Decision Process
```text
Existing Project
│
▼
Check requirements.txt
│
▼
Identify library versions
│
▼
Check supported Python versions
│
▼
Install compatible Python version
│
▼
Create venv
│
▼
Install libraries
│
▼
Run project
```

### For New Projects
For new development, use the latest stable Python version whenever possible.
Example:
```text
Python 3.13.x
```
Benefits:
- Performance improvements
- Security updates
- New language features
- Longer support lifecycle

### For Legacy Projects
Do not upgrade Python blindly.
Examples:
```text
Project A
Libraries from 2021
→ Python 3.9 or 3.10
```
```text
Project B
Libraries from 2023
→ Python 3.11
```
```text
Project C
Latest libraries
→ Python 3.13
```

## How do we "run" Python?

→ Python is a compiler/interpreter

&emsp;→ reads in a chunk of code (your program)

&emsp;&emsp;&emsp;&emsp;→ compiles and runs it

&emsp;&emsp;&emsp;&emsp;→ output is sent to your screen (console)

→ Python can do this in two ways

&emsp;→ `interactive` mode

&emsp;&emsp;&emsp;&emsp;→ you type a Python line/block of code and execute it immediately

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;→ any output is immediately displayed

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;→ continue typing/running code one line/block at a time

&emsp;&emsp;&emsp;&emsp;→ REPL (read-eval-print-loop)

&emsp;→ `script` mode

&emsp;&emsp;&emsp;&emsp;→ write all your code in files first

&emsp;&emsp;&emsp;&emsp;→ then execute all this code using command line

# `working with Jupyter Notebooks`

   - Activate the virtual environment first, then install libraries from any folder path in the command prompt.If the venv is activated, pip installs inside that venv, regardless of folder path you are in
   - With venv active, type: (This installs the classic Jupyter Notebook application - Only Jupyter Notebook (lighter, sufficient for most users)
     ```
     pip install notebook
     ```
     ```
     pip install jupyterlab
     ```
     If you want the newer Jupyter ecosystem (Notebook, Lab, console, etc.), you can install - Meta-package that installs Notebook plus several other Jupyter tools:
     ```
     pip install jupyter
     ```
     Install Python execution engine (kernel) that actually runs your code.
     ```
     pip install ipykernel
     ```
   - Start it:
     ```
     jupyter notebook
     ```
     ```
     jupyter lab
     ```
     Browser opens. Click "New > Python 3" for a notebook. Test: Type `print("Hi!")` in a cell, hit Shift+Enter.
   - In jupyter notebook `view tab` > `open in jupyter lab` to open the notebook in jupyter lab
     
   - To stop a running Jupyter Notebook, go to the Command Prompt where it was launched and press:
     ```
     Ctrl + C
     ```
     You will see:
     ```
     Shutdown this notebook server (y/[n])?
     ```
     Type:
     ```
     y
     ```
     

   - [Interface tutorial](https://www.youtube.com/watch?v=uclhm_l1cSc)

Change the Default Python in VS Code

1. Press `Ctrl + Shift + P`
2. Select **Python: Select Interpreter**
3. Choose **Python 3.11**
4. Restart the terminal/notebook kernel.
5. For notebooks, select the **Python 3.11** kernel.
   
VS Code will use Python 3.11 for that workspace.

**Note:** You do not need to navigate to the Python installation folder.
