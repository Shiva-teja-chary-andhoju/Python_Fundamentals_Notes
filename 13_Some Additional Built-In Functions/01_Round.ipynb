# Rounding

→ `round()` is a built-in function that can be used to round floats

&emsp;→ uses `banker's rounding`

&emsp;&emsp;&emsp;→ also called `round half to even`

| | |
|---|---|
| → rounds away from zero | `1.8` → `2`<br>`-1.8` → `-2` |
| → ties round to closest even digit | `1.5` → `2`<br>`2.5` → `2` |

&emsp;&emsp;&emsp;→ good choice to eliminate various `biases`

→ use `round()` to round to an integer

```python
round(1.8)
```

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;→ `2`

```python
round(-1.8)
```

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;→ -`2`

```python
round(1.5)
```

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;→ `2`

```python
round(2.5)
```

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;→ `2`

→ can also use `round()` to round to closest multiple of 1/10

    round(value, exponent)

`exponent` is used to specify what power of 1/10 to round to

→ let's look at it mathematically first (i.e. without worrying about float representations)

    round(x, 1)   → rounds to nearest 0.1   (10^-1)

    round(x, 2)   → rounds to nearest 0.01  (10^-2)

    round(x, -1)  → rounds to nearest 10    (10^1)

    round(x, -2)  → rounds to nearest 100   (10^2)

round to closest multiple of:

| Expression | Multiple Of | Result |
|------------|-------------|---------|
| `round(127.1892, 3)` | `10^-3` → `0.001` | `127.189` |
| `round(127.1892, 2)` | `10^-2` → `0.01` | `127.19` |
| `round(127.1892, 1)` | `10^-1` → `0.1` | `127.2` |
| `round(127.1892, 0)` | `10^0` → `1` | `127.0` |
| `round(127.1892, -1)` | `10^1` → `10` | `130.0` |
| `round(127.1892, -2)` | `10^2` → `100` | `100.0` |
| `round(127.1892, -3)` | `10^3` → `1000` | `0.0` |

## Rounding Ties in Floats

→ technically rounds to closest number that ends with an even digit

&emsp;&emsp;`round(0.125, 2)` → `0.12`

&emsp;→ so why this?

| | |
|---|---|
| `round(0.325, 2)` → `0.33` | why not `0.32`? |

&emsp;&emsp;→ remember floats do not have (in general) an exact representation!

&emsp;&emsp;&emsp;`0.325` → `0.325000000000000011102230246252`

&emsp;&emsp;&emsp;→ so this is **not** a tie!

# Coding

### round

We can use the `round()` built-in function to round floats and integers.

By default, `round()` will round to the closest integer (using banker's rounding):

````python
round(0.325)
````

````python
round(0.875)
````

And for ties:

````python
round(13.5)
````

````python
round(12.5)
````

As you can see it rounds to the closest **even** integer.

We can pass a second argument to `round()` to specify what multiple of 1/10 we want to round to.

For example, we can round to the closest multiple of `0.1` this way:

````python
round(0.125, 1)
````

and to the closest multiple of `0.01`:

````python
round(0.125, 2)
````

But that second argument can be negative, and that allows us to round to the closest multiple of `10`, `100`, etc...

For example:

````python
round(123456, -1)
````

````python
round(123456, -2)
````

````python
round(123456, -3)
````

Again, ties will go the closest number ending in an even digit:

````python
round(1235, -1)
````

````python
round(1245, -1)
````

Let's go back to floats for a bit now. The problem is that floats do not have exact representations (at least the majority of them).

A number such as `0.125` does have an exact representation, so if we round to the closest multiple of `0.01`:

````python
round(0.125, 2)
````

we see that the rounding rounded down to `1.2` - since `1.3` ends with an odd digit, `1.2` was picked.

But wee may sometimes observe odd behavior, where this is not happening!

````python
round(0.325, 2)
````

We woud have expected the answer to be `0.32`, not `0.33`. An in a prefect world, that would be the case - but `0.325` does not have an exact representation as a float:

````python
format(0.325, '.20f')
````

As you can see, this number is not **exactly** `0.325`, but rather slightly higher. This means that when we round this number we do **not** actually have a tie - it simply rounds to the closest multiple of `0.01`, which is `0.33`.
