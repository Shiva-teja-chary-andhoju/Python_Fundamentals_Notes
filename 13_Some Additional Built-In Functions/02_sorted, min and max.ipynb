# `sorted`, `min` and `max`

## Sorting Numbers

→ numbers have a `natural` sort order

→ they can be sorted `ascending` or `descending` by that `sort order`

→ `sorted` is a built-in function that can be used to sort a collection of numbers

&emsp;&emsp;→ single positional argument: an `iterable` containing the numbers

&emsp;&emsp;→ by default, it sorts in `ascending` order

&emsp;&emsp;&emsp;→ keyword-only argument to `reverse` the sort order

&emsp;&emsp;&emsp;&emsp;&emsp;→ default is `False`  → sorts` ascending `

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;→ specify `reverse=True`  → sorts `descending `

&emsp;&emsp;→ always returns a new `list`

&emsp;&emsp;&emsp;→ original iterable is `not mutated`

```python
t = (1, 10, 2, 9, 3, 8)
```

```python
sorted(t)
```

&emsp;→ `[1, 2, 3, 8, 9, 10]`

```python
sorted(t, reverse=True)
```

&emsp;→ `[10, 9, 8, 3, 2, 1]`

## Sorting Strings

Numbers have a natural sort order

Strings also have a natural sort order in Python

&emsp;→ `lexicographic` order

&emsp;&emsp;&emsp;→ `dictionary` order, `alphabetical` order


 **BEWARE**  The characters `a` and `A` are not the same 

→ Python assigns a numerical character code (the unicode character

code) to each character in a string

| | |
|---|---|
| `A` → `65` | `Z` → `90` |
| `a` → `97` | `z` → `122` |

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;→`'A' < 'Z' < 'a' < 'z'`





→ so Python will use "alphabetical" sorting, but upper case letters will be

sorted before their equivalent lower case versions

→ natural sort order of string is case sensitive

```python
sorted(['Boy', 'baby'])
```

→ `['Boy', 'baby']`

 (`ascending` sort)

## `min` and `max`

→ closely related to sorting

&emsp;&emsp;→ to find the `minimum` of a collection

&emsp;&emsp;&emsp;→ `sort` the collection (`by` something) in `ascending` order

&emsp;&emsp;&emsp;→ pick the `first` element

&emsp;&emsp;→ to find the maximum of a collection

&emsp;&emsp;&emsp;→ `sort` the collection (`by` something) in `descending` order

&emsp;&emsp;&emsp;→ pick the `first` element

(or you could sort in the other direction in both cases and pick the last element)

```python
min([1, 10, 2, 9, 8])
```

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;→ `1`

```python
max([1, 10, 2, 9, 8])
```

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;→ `10`

```python
min([])
```

&emsp;&emsp;&emsp;&emsp;&emsp;→ `ValueError` exception

&emsp;→ can specify a `default` value to return if the iterable is empty

&emsp;&emsp;&emsp;&emsp;→ keyword-only argument

&emsp;&emsp;&emsp;&emsp;`min([], default=0)` → `0`

→ can also use an arbitrary number of positional arguments instead

&emsp;`min(1, 10, 2, 9, 3, 8)` → `1`

&emsp;`max(1, 10, 2, 9, 3, 8)` → `10`

→ we'll come back to `min` and `max` when we look at sorting again later in this course

### Sorted, min and max

In this video we are only going to look at numbers and strings, using the natural ordering that exists for those two types.

We'll come back to `sorted`, `min` and `max` again later.

Let's start with an iterable of numbers - it could be a tuple, a list, or even a set:

````python
l = [1, 10, 2, 9, 3, 8]
t = (1, 10, 2, 9, 3, 8)
````

````python
sorted(l)
````

````python
sorted(t)
````

You'll notice that in each case we get a `list` back.

Note also that this is not an in-place sort - in the sense that the original iterable was not modified in any way:

````python
l
````

````python
t
````

The default sort order is ascending (from smallest to largest), but we can change this to descending by specifying a key-word-only argument:

````python
sorted(l, reverse=True)
````

Strings also have a natural ordering, based on the character code of each character in the string. But this means string sorts are case sensitive:

````python
sorted(('a', 'z', 'b', 'y', 'c', 'x'))
````

But this may seem odd:

````python
sorted({'a', 'A', 'b', 'B', 'x', 'X'})
````

This is because the character code for `A` is smaller than the character code for `a`:

````python
'A' < 'a'
````

and in fact:

````python
'Z' < 'a'
````

So, 

````python
sorted(['atom', 'apple', 'Zebra'])
````

We'll come back to sorting later and see how we can change this default behavior.

The `min` and `max` functions are very much related to sorting.

To find the smallest element in a list of numbers we could do this:

````python
l = [1, 10, 2, 9, 3, 8]
````

````python
sorted_ascending = sorted(l)
````

````python
smallest_element = sorted_ascending[0]
````

````python
smallest_element
````

Instead of doing this, we can use `min`:

````python
min(l)
````

If we pas an empty iterable to the `min` or `max` functions, we will get an exception - it is not possible to find the smallest element of an empty collection after all!

````python
try:
    min([])
except ValueError as ex:
    print('ValueError:', ex)
````

If we want, we can specify a `default` value, a keyword-only argument to use as the min/max if the iterable is empty:

````python
min([], default=0)
````

The `min` function takes a single iterable as a positional argument, but it can also handle *multiple* positional arguments instead:

````python
min(1, 10, 2, 9, 3, 8)
````

Of course, using this variant means that using a `default` makes no sense:

````python
min(1, 10, default=0)
````

`max` works exactly the same way.
