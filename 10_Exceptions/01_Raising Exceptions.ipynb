# Raising Exceptions

## What are exceptions?

→ exceptions are special events that happen when something out of the

ordinary happens while our code is running

→ an exception is generally unexpected behavior

&emsp;&emsp;→ but not always

&emsp;&emsp;→ it may be something we expect to happen from time to time

&emsp;&emsp;&emsp;&emsp;→ we can deal with it and continue running our code

→ so an exception is not necessarily an error

→ but unhandled exceptions will cause our program to terminate

## Terminology

| | |
|---|---|
| `exception` | → a special type of `object` in Python |
| `raising` | → starting an exception `event flow` |
| `exception handling` | → `interacting` with an exception flow in some manner |
| `unhandled exception` | → an exception flow that is `not handled` by our code<br>→ generally results in our program terminating abruptly |

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;

## Exception Hierarchy

→ Python exceptions form a hierarchy

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;(we'll cover what that means precisely when we look at Object Oriented Programming - OOP)

&emsp;&emsp;https://docs.python.org/3/library/exceptions.html#exception-hierarchy

→ basically means that exceptions can be classes sub-divided into sub- exceptions that are more specific

→ for example a broad exception might be `LookupError`

&emsp;&emsp;→ more specifically it could be an `IndexError` or a `KeyError`

&emsp;&emsp;&emsp;→ both of these are categorized more broadly as a `LookupError`

&emsp;&emsp;&emsp;&emsp;→ can choose to handle `IndexError` specifically

&emsp;&emsp;&emsp;&emsp;→ or `LookupError` more broadly

## Exception Hierarchy

this means that if an exception object is an `IndexError` exception

&emsp;→ it is also a `LookupError` exception

&emsp;→ and it is also an `Exception` exception

&emsp;&emsp;&emsp;&emsp;(most exceptions we work with are classified as `Exception` types)

→ confused?  think of it this way… 

Say we have these classes of objects:

```text
                 College Person
                 ──────────────

           Student      Staff Member

                   ┌──────────┴──────────┐
```

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;→ a Teacher is also Staff Member

## EAFP vs LBYL

→ when we think something unexpected may go wrong in our code

&emsp;&emsp;→ figure out if something is going to wrong before we do it

| | |
|---|---|
| → LBYL | **L**ook **B**efore **Y**ou **L**eap |

&emsp;&emsp;→ just do it, and handle the exception if it occurs

| | |
|---|---|
| → EAFP | **E**asier to **A**sk **F**orgiveness than **P**ermission |

<br>

| | |
|---|---|
| generally in Python | → follow EAFP<br>→ exception handling |

## Why EAFP?

Something that is exceptional should be `infrequent`

&emsp;&emsp;→ if we are dividing two integers in a loop that repeats 1,000 times

&emsp;&emsp;&emsp;→ out of every 1,000 times we run, we expect division by zero to occur 5 times

&emsp;&emsp;&emsp;&emsp;&emsp;LBYL → test that divisor is non-zero 1,000 times

&emsp;&emsp;&emsp;&emsp;&emsp;EAFP → just do it, and handle the division by zero error 5 times

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;→ often more efficient

→ also trying to fully determine if something is `going` to go wrong is a `lot harder` to write than just handling things when they `do` go wrong

## Exception Handling Flow

→ an exception occurs

&emsp;&emsp;→ an exception object is created

&emsp;&emsp;→ an exception Clow is started

&emsp;&emsp;→ we do nothing about it

&emsp;&emsp;&emsp;&emsp;→ program terminates

<br>

&emsp;&emsp;→ we intercept the exception flow

&emsp;&emsp;&emsp;&emsp;→ try to handle the exception in some sense, if possible

&emsp;&emsp;&emsp;&emsp;→ then

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;→ resume running program uninterrupted

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;→ or, let the exception resume

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;→ or, start a new exception flow

# Raising Exceptions


→ often we want to start an exception flow ourselves

&emsp;→ called `raising` an exception

→ an exception object is associated with an exception flow

&emsp;→ we create a new exception object

&emsp;→ we raise the exception object

→ doing this is most useful when we create functions

&emsp;→ we'll see this later

&emsp;&emsp;→ for now, we'll just learn how to raise an exception

**Example**

→ create an exception object using one of Python's built-in exceptions

```python
ex = ValueError()
```

&emsp;→ usually we include a custom exception message

```python
ex = ValueError('Name must be at least 5 characters long.')
```

→ we raise the exception, starting an exception flow

```python
raise ex
```

→ often do both in one step

```python
raise ValueError('custom message')
```


→ raising an exception ourselves results in the same exception flow that Python does when it raises some exception

&emsp;→ we can choose to handle the exception

&emsp;→ if we don't handle the exception, program terminates

We have seen exception flows before:

````python
a = 1
b = 0
result = 1 / 0
print(result)
````

As you can see, a `ZeroDivisionError` exception was **raised** by Python.

If this had been a normal Python program, it would have terminated.

Here we are running inside a Jupyter notebook, so the notebook itself does not terminate - instead Jupyter "handles" the exception by printing out the exception and stops processing the remainder of the cell. (the `print` statement never ran).

We can create our own exception objects using Python's built-in exceptions (we can also create our own, which we'll see later).

````python
ex = ValueError('Name must be at least 5 characters long.')
````

Notice that creating an exception object did **not** start an exception flow! We just created an object - nothing more.

````python
type(ex)
````

The representation of the exception (what Jupyter prints out) is:

````python
ex
````

We can also get that representation directly by using the `repr()` function:

````python
repr(ex)
````

This is different than the string representation, which is used by `print` and can also be recovered using the `str` function:

````python
print(ex)
````

````python
str(ex)
````

So we often use the `str` function to get to a string containing just the error message.

If we want to start an exception flow, we have to `raise` the exception:

````python
raise ex
````

As you can see, Jupyter intercepted the exception and handled it for us. If this had been a normal program, it would have terminated.

Let's try this with an input value for a name, raising an exception if user does not enter a name that is at least 5 characters long:

````python
name = input('Enter name (5 chars min): ')

if len(name) < 5:
    raise ValueError(f'{name} is not 5 characters or more...')
    
print(f'Hello {name}!')
````

````python
name = input('Enter name (5 chars min): ')

if len(name) < 5:
    raise ValueError(f'{name} is not 5 characters or more...')
    
print(f'Hello {name}!')
````

So now we have a way to raise exceptions when we need to, but we still need to learn how to handle exceptions (exception flows, or raised exceptions, to be exact).

As we discussed in the lecture, Python has a class hierarchy for exceptions.

For example, we have `KeyError` and `IndexError` that are both also `LookupError` exceptions, which itself is also an `Exception` exception.

````python
issubclass(KeyError, LookupError)
````

````python
issubclass(KeyError, Exception)
````

````python
issubclass(LookupError, Exception)
````

This means that instances of `KeyError` are also considered to be instances of `LookupError` as well as `Exception`:

````python
ex = KeyError('some message')
````

````python
type(ex)
````

````python
isinstance(ex, KeyError)
````

````python
isinstance(ex, LookupError)
````

````python
isinstance(ex, Exception)
````

But `KeyError` and `IndexError` are not subclasses of each other - they are considered "siblings" of the same parent class (`LookupError`):

````python
issubclass(KeyError, IndexError)
````

````python
isinstance(ex, IndexError)
````

The fact that Python exceptions form this type of hierarchy becomes very relevant when it comes to exception handling.
